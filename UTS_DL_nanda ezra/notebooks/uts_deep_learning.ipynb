# UTS Deep Learning — jalankan di Google Colab

Notebook ini menjalankan tiga tugas (klasifikasi cacat, prediksi RUL, clustering) di Colab. Siapkan uts_dl_solution_colab_source.zip yang disertakan bersama notebook ini. Dataset dicari dari folder Google Drive yang dibagikan kepada Anda.

Langkah: unggah notebook ini ke Colab, pilih Runtime → Change runtime type → T4 GPU jika tersedia, lalu pilih Runtime → Run all. Saat diminta, unggah ZIP kode. Izinkan Colab mengakses Google Drive. Proses penuh dapat memakan waktu lama; jangan tutup tab sebelum selesai.

File hasil akan tersimpan di MyDrive/uts_dl_colab/hasil_uts_dl. Dataset tidak diunggah kembali dan tidak masuk ke GitHub.

```python
from google.colab import drive, auth, files
from pathlib import Path
import os, shutil, zipfile, subprocess, sys, json, time

drive.mount('/content/drive')
auth.authenticate_user()

PROJECT = Path('/content/uts_dl_solution')
OUTPUT = Path('/content/drive/MyDrive/uts_dl_colab/hasil_uts_dl')
OUTPUT.mkdir(parents=True, exist_ok=True)
print('Drive terpasang. Hasil akan disimpan ke:', OUTPUT)
```

## 1. Unggah source code proyek

Pilih file uts_dl_solution_colab_source.zip yang Anda unduh bersama notebook. ZIP ini hanya berisi kode program dan requirements, tidak berisi dataset atau model besar.

```python
if not PROJECT.exists() or not (PROJECT / 'kode' / 'pipeline.py').exists():
    uploaded = files.upload()
    zip_names = [name for name in uploaded if name.lower().endswith('.zip')]
    if not zip_names:
        raise RuntimeError('Pilih file uts_dl_solution_colab_source.zip terlebih dahulu.')
    with zipfile.ZipFile(zip_names[0]) as z:
        z.extractall(PROJECT)
    if not (PROJECT / 'kode' / 'pipeline.py').exists():
        raise RuntimeError('Isi ZIP tidak sesuai. Pastikan memakai uts_dl_solution_colab_source.zip.')
print('Kode proyek siap:', PROJECT)
```

## 2. Temukan dataset pada folder Drive yang dibagikan

Notebook memakai Drive API setelah Anda mengizinkan akses. Ia mencari data_uts_dl.zip di folder tersebut dan subfoldernya. Jika Drive API tidak menemukan arsip, pastikan folder dibagikan ke akun Google yang sama dengan Colab.

```python
from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseDownload
import io
from IPython.display import display, Markdown

SHARED_FOLDER_ID = '1B1ACbd8Sdif_M6a6YOs1_3q_Kg597uRM'
drive_api = build('drive', 'v3')
queue = [SHARED_FOLDER_ID]
visited = set()
found = None
while queue and found is None:
    folder_id = queue.pop(0)
    if folder_id in visited:
        continue
    visited.add(folder_id)
    page_token = None
    while True:
        result = drive_api.files().list(
            q=f"'{folder_id}' in parents and trashed = false",
            fields='nextPageToken,files(id,name,mimeType,size)',
            pageToken=page_token,
            includeItemsFromAllDrives=True,
            supportsAllDrives=True,
            pageSize=1000,
        ).execute()
        for item in result.get('files', []):
            if item['mimeType'] == 'application/vnd.google-apps.folder':
                queue.append(item['id'])
            elif item['name'].lower() == 'data_uts_dl.zip':
                found = item
                break
        if found or not result.get('nextPageToken'):
            break
        page_token = result['nextPageToken']

if found is None:
    raise FileNotFoundError(
        'Tidak menemukan data_uts_dl.zip pada folder Drive. Pastikan nama arsip tepat '
        'dan akun Colab memiliki akses ke folder yang dibagikan.'
    )
DATA_ZIP = Path('/content/data_uts_dl.zip')
request = drive_api.files().get_media(fileId=found['id'], supportsAllDrives=True)
with io.FileIO(DATA_ZIP, 'wb') as fh:
    downloader = MediaIoBaseDownload(fh, request)
    done = False
    while not done:
        status, done = downloader.next_chunk()
        if status:
            print(f'Download dataset: {int(status.progress() * 100)}%')
print(f"Dataset ditemukan: {found['name']} ({int(found.get('size', 0)) / 1e6:.1f} MB)")
with zipfile.ZipFile(DATA_ZIP) as dataset_zip:
    display(Markdown(dataset_zip.read('kamus_data.md').decode('utf-8')))
```

## 3. Pasang dependensi dan jalankan seluruh pipeline

Colab sudah menyediakan PyTorch dan Jupyter. Sel ini memasang paket lain yang diperlukan, lalu menjalankan konfigurasi penuh proyek: 10 epoch klasifikasi, 12 epoch RUL, 20 epoch clustering, dan tiga seed (7, 21, 42).

```python
req = (PROJECT / 'kode' / 'requirements.txt').read_text().splitlines()
install = [line for line in req if line.strip() and not line.lower().startswith(('torch', 'jupyter'))]
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', *install])

cmd = [
    sys.executable, '-m', 'kode.pipeline',
    '--data', str(DATA_ZIP), '--out', str(OUTPUT),
    '--epochs-class', '10', '--epochs-rul', '12', '--epochs-cluster', '20',
    '--seeds', '7', '21', '42',
]
print('Menjalankan pipeline pada:', 'GPU' if __import__('torch').cuda.is_available() else 'CPU')
subprocess.run(cmd, cwd=PROJECT, check=True)
print('Pipeline selesai. Semua hasil utama ada di:', OUTPUT)
```
